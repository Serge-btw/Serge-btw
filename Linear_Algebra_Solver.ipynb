```python
import matplotlib.pyplot as plt

# 1. Mock Instagram Content Performance Data
content_types = ['Reels (Tech Tutorials)', 'Carousel Posts', 'Standard Photos', 'IG Stories']
average_reach = [4200, 2400, 1100, 650]  # Average unique accounts reached

# 2. Design and customize the performance graph
plt.figure(figsize=(8, 5))
colors = ['#833AB4', '#C13584', '#E1306C', '#F77737'] # Canonical Instagram theme colors

# 3. Generate the horizontal bar chart
plt.barh(content_types, average_reach, color=colors, edgecolor='black')

# 4. Add clear labels and visual titles
plt.title('Instagram Content Performance Analysis', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Average Reach (Unique Accounts)', fontsize=11, labelpad=10)
plt.ylabel('Content Formats', fontsize=11)
plt.grid(axis='x', linestyle='--', alpha=0.5) # Add a subtle background tracking grid

# 5. Render the chart
plt.show()
```
