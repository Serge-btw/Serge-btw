```python
import numpy as np

# 1. Define the matrix of coefficients (the numbers in front of x and y)
A = np.array([[2, 1],
              [1, -3]])

# 2. Define the matrix of constants (the answers on the right side)
B = np.array([8, -3])

# 3. Use NumPy's linear algebra solver to compute x and y automatically
solution = np.linalg.solve(A, B)

# 4. Print the final calculated values
print(f"Calculated Solution: x = {solution[0]}, y = {solution[1]}")
```
